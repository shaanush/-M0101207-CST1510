### D13. Read a file line by line and split each line

`records_sample.txt` (in this `LAB` folder) has one record per line, like `R-001,87,100`.

1. Open it with `with` and loop over it line by line.
2. For each line, use `.strip()`, then `.split(",")` into `label`, `value`, `limit`.
3. Convert `value` and `limit` with `float()`, and print all three.

**Expected output:**
```
R-001 87.0 100.0
R-002 120.0 100.0
R-003 45.0 100.0
```

````python
# D13
with open("records_sample.txt" , "r") as f:
    for line in f:
        line = line.strip()
        part = line.split(",")
        label = part[0]
        value = float(part[1])
        limit = float(part[2])
        print(label,value,limit)
````

### C1. Load records safely from a file

Write `load_records(path)` with a one-line docstring. It reads `label,value,limit` lines from a file
and returns a **list of dicts**. If the file does not exist, it prints a clear message and returns an
empty list instead of crashing.

Then load `records_sample.txt` and print each record's label and status, as in D16.

**Test it:**

| Call | Expected output |
|---|---|
| `load_records("records_sample.txt")` | the same three lines as D16 |
| `load_records("nope.txt")` | a clear message, then `[]` |

````python
# C1
def load_records(path): # path is a variable that holds the file name "sample.txt"
    """ this function will use path to store file name and then open it if it exists"""
    records = []        #use to store dict
    try:
        with open(path,"r") as f: #in path it stores the file NAME then after it open that specific file name when we open it it is now store in F
            for line in f:      # it read line inside f....f means the opening file 
                line = line.strip()   # we use this to remove like \n and all to make everything clean
                if not line:
                    continue
                parts = line.split(",")
                if len(parts) != 3:
                    continue
                record = {
                "label": parts[0],
                "value": parts[1],
                "limit": parts[2],
                }
                records.append(record)
        return records
    except FileNotFoundError:
        print("File not found")
        return[]
print(load_records("records_sample.txt"))
print(load_records("nope.txt"))








````
